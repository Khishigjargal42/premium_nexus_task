# Business Insights & Recommendations

This notebook summarizes the key findings from sales, customer, product,
inventory, and survey analysis and translates them into actionable business recommendations.

## 1. Prioritize Inventory Monitoring for High-Value Products

### Finding
Pasta Box generated the highest product revenue at 8.0M, while showing
28.4% low-stock days and 1.8% OOS days. Cheesecake Slice and Beef Onigiri
also showed relatively high inventory pressure.

### Recommendation
Prioritize inventory monitoring for high-revenue products with frequent
LOW or OOS status, particularly Pasta Box, Cheesecake Slice, and Beef Onigiri.

### Expected Impact
Reduce availability risk for high-value products and support more consistent sales.

### Limitation
The available data does not directly measure lost sales caused by stockouts.
Therefore, the analysis identifies inventory risk rather than estimated revenue loss.

## 2. Strengthen Customer Retention

### Finding
All 120 identified customers were repeat customers, with an average of
35.6 transactions per customer. Morning commuter and Healthy routine
segments showed the highest revenue per customer.

### Recommendation
Use targeted loyalty and personalized promotions to retain high-frequency
customer segments, particularly Morning commuter and Healthy routine.

### Expected Impact
Maintain repeat purchase behavior and increase customer lifetime value.

## 3. Improve Product Quality

### Finding
Customers who answered "yes" to buying again had an average quality score
of 7.17, compared with 5.99 among customers who answered "no".

Products with relatively low quality scores included Iced Coffee,
Protein Bar, and Chicken Sandwich.

### Recommendation
Review product quality for lower-scoring products and monitor changes in
quality scores and repurchase intent after improvements.

### Expected Impact
Improve customer experience and potentially strengthen repurchase intent.

### Limitation
The analysis shows an association between quality scores and repurchase intent,
but does not establish causation.

## 4. Optimize Product Mix

### Finding
ReadyToEat generated the largest category revenue share at 34.6%.
Beverage had the highest unit volume but contributed only 14.2% of revenue.

### Recommendation
Protect availability of ReadyToEat products as a core revenue category.
For Beverage products, test cross-selling or bundle strategies to increase
revenue per transaction.

### Expected Impact
Improve revenue contribution from existing customer traffic and optimize
the product mix.
