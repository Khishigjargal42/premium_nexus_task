```python
checks = {
    "Sales → Customers": (
        sales["customer_id"].dropna(),
        customers["customer_id"]
    ),
    "Sales → Branches": (
        sales["branch_id"],
        branches["branch_id"]
    ),
    "Sales Line → Sales": (
        sales_line["transaction_id"],
        sales["transaction_id"]
    ),
    "Sales Line → Products": (
        sales_line["product_code"],
        products["product_code"]
    ),
    "Inventory → Branches": (
        inventory["branch_id"],
        branches["branch_id"]
    ),
    "Inventory → Products": (
        inventory["product_code"],
        products["product_code"]
    ),
    "Survey → Branches": (
        survey["branch_id"],
        branches["branch_id"]
    ),
    "Survey → Products": (
        survey["product_code"],
        products["product_code"]
    ),
}

for name, (child, parent) in checks.items():
    unmatched = (~child.isin(parent)).sum()
    print(f"{name}: {unmatched} unmatched")
```

```python
transaction_totals = (
    sales_line
    .groupby("transaction_id")["line_amount"]
    .sum()
    .rename("calculated_amount")
)

sales_check = sales.merge(
    transaction_totals,
    on="transaction_id",
    how="left"
)

sales_check["difference"] = (
    sales_check["basket_amount"] -
    sales_check["calculated_amount"]
)

print("Transactions:", len(sales_check))
print("Exact matches:", (sales_check["difference"] == 0).sum())
print("Non-zero differences:", (sales_check["difference"] != 0).sum())
```

```python
sales_line_check = sales_line.copy()

sales_line_check["calculated_amount"] = (
    sales_line_check["qty"] * sales_line_check["unit_price"]
    - sales_line_check["discount_amt"]
)

sales_line_check["difference"] = (
    sales_line_check["line_amount"]
    - sales_line_check["calculated_amount"]
)

print("Sales lines:", len(sales_line_check))
print("Exact matches:", (sales_line_check["difference"] == 0).sum())
print("Non-zero differences:", (sales_line_check["difference"] != 0).sum())
```

```python
date_columns = {
    "Sales - transaction_ts": sales["transaction_ts"],
    "Sales - sales_date": sales["sales_date"],
    "Customers - join_date": customers["join_date"],
    "Inventory - inventory_date": inventory["inventory_date"],
    "Survey - feedback_date": survey["feedback_date"],
}

for name, col in date_columns.items():
    parsed = pd.to_datetime(col, errors="coerce")
    invalid = parsed.isna().sum()
    
    print(f"{name}: {invalid} invalid dates")
```
